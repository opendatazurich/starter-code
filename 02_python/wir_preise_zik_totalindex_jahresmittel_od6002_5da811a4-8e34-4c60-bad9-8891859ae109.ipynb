## Description

Konsumentenpreisindex für die Stadt Zürich mit verschiedenen Basisjahren


**Zürcher Index der Konsumentenpreise (ZIK)**

Der ZIK misst die Teuerung der Konsumgüter und Dienstleistungen in der Stadt Zürich. Der ZIK zeigt, um wie viel die Konsumgüter und Dienstleistungen gegenüber dem Vormonat, dem Vorjahr oder jedem anderen früheren Zeitpunkt teurer geworden sind.

[]()

**Basis**

Die Basis eines Index ist der feste Ausgangs- und Bezugswert (Indexstand = 100), an dem alle späteren Veränderungen gemessen werden und der den Startpunkt für Vergleiche über die Zeit bildet. Sie setzt sich immer aus Basisjahr und Basismonat zusammen. Seit dem Jahr 2000 wird der ZIK alle 5 Jahre revidiert, wobei der Basismonat seit der Revision 2005 einheitlich auf Dezember normiert ist.

Zum Revisionszeitpunkt wird der Warenkorb angepasst und der Index neu auf 100 gesetzt. Da sich mit dem neuen Warenkorb auch die Berechnungsgrundlage ändert, sind Indexstände vor und nach einer Revision nicht direkt vergleichbar. Die Rücksetzung auf 100 markiert deshalb den eindeutigen Ausgangspunkt der neuen Zeitreihe. Frühere Zeitreihen mit älteren Basen bleiben davon unberührt und weiterhin verfügbar. Weitere Informationen zum Warenkorb und zur aktuellen Gewichtung der einzelnen Positionen finden Sie unter:  [**Warenkorb Stadt Zürich**](https://www.stadt-zuerich.ch/de/politik-und-verwaltung/statistik-und-daten/daten/wirtschaft/zuercher-index-der-konsumentenpreise/warenkorb.html).

[]()

**Indexstand**

Der Indexstand wird in Punkten angegeben und ist immer im Verhältnis zur jeweiligen Basis zu lesen. Beispiel: Steht der Index (Basis 2020 = 100) heute bei 105 Punkten, sind die Konsumgüter und Dienstleistungen seit 2020 um 5 % teurer geworden (siehe Berechnung Veränderungsrate). Anhand der Indexstände lässt sich die Preisentwicklung über die Zeit verfolgen, weshalb der Konsumentenpreisindex zu den wichtigsten Wirtschafts- sowie Inflationsindikatoren zählt.

[]()

**Veränderungsrate**

Die Veränderungsrate (in %) zwischen zwei beliebigen Zeitpunkten wird anhand zweier Indexstände ermittelt, die auf derselben Basis berechnet wurden. Im Gegensatz zum Indexstand ist die Veränderungsrate unabhängig von der aktuellen Basis und bleibt auch über einen Basiswechsel hinweg direkt vergleichbar. Die Veränderungsrate wird wie folgt berechnet: Veränderungsrate (%) = (Indexstand neu - Indexstand alt) / Indexstand alt × 100

```python
# helper function for reading datasets with proper separator
def get_dataset(url):
    """
    Return pandas df if url is parquet or csv file. Return None if not.
    """
    extension = url.rsplit(".",1)[-1]
    if extension == 'parquet':
        data = pd.read_parquet(url, storage_options=STORAGE_OPTIONS)
    elif extension == 'csv':
        data = pd.read_csv(url, sep=",", on_bad_lines='warn', encoding_errors='ignore', low_memory=False, storage_options=STORAGE_OPTIONS)
        # if dataframe only has one column or less the data is not comma separated, use ";" instead
        if data.shape[1] <= 1:
            data = pd.read_csv(url, sep=';', on_bad_lines='warn', encoding_errors='ignore', low_memory=False, storage_options=STORAGE_OPTIONS)
            if data.shape[1] <= 1:
                print("The data wasn't imported properly. Very likely the correct separator couldn't be found.\nPlease check the dataset manually and adjust the code.")
    else:
        print("Cannot load data! Please provide an url with csv or parquet extension.")
        data = None
    return data
```
